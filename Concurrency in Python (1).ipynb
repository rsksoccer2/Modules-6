```python
import datetime

# 13.1 Write the current date as a string to today.txt
today = datetime.date.today()

with open("today.txt", "w") as file:
    file.write(str(today))

# 13.2 Read today.txt into today_string
with open("today.txt", "r") as file:
    today_string = file.read()

print("Today:", today_string)

# 13.3 Parse the date from today_string
parsed_date = datetime.datetime.strptime(today_string, "%Y-%m-%d").date()

print("Parsed date:", parsed_date)
```

```python
import multiprocessing
import random
import time
from datetime import datetime

def print_time():
    # Random wait 0-1 sec
    wait = random.random()
    time.sleep(wait)
    now = datetime.now()
    print(f"{multiprocessing.current_process().name} waited {wait:.2f}s - {now.strftime('%H:%M:%S.%f')}")

if __name__ == '__main__':
    processes = []

    for i in range(3):
        p = multiprocessing.Process(target=print_time)
        processes.append(p)
        p.start()

    for p in processes:
        p.join()

    print("Done.")
```
